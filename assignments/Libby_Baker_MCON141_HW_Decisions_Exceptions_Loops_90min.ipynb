## 1. Nested decision: library access (10 minutes)

Write a function named `check_library_access()` with no arguments and no return. Ask for a student's age and whether they have a library card (`yes` or `no`). Use a **nested** `if` structure to print one appropriate message:

- Age 18 or older with a card: may borrow books.
- Age 18 or older without a card: may apply for a card.
- Under 18 with a card: may borrow with a guardian's permission.
- Under 18 without a card: must bring a guardian to apply.

Call the function below.

```python
# Write your function here

def check_library_access():
  # gather inputs for card and age
  card = input("Do you have a library card? yes/no").strip().lower()
  age = int(input("How old are you?"))

  if age >=18:#nested descision if over 18
    if card == "yes": #has a card
      print("You may borrow books")
    else: # doesn't have a card
      print("You may apply for a card")
  else: # nested descision for if under 18
    if card == "yes": # has card
      print("You may borrow with a guardian's permission")
    else: # doesn't have card, is under 18
      print("You must bring a guardian to apply")

check_library_access()
```

## 4. Raise a meaningful error (12 minutes)

Write `get_course_level()` with no arguments and no return. Ask for a course level as an integer.

- Use `try` to convert the input.
- If the number is not from 100 through 499 inclusive, use `raise ValueError(...)` with a useful message.
- Use `except ValueError as e` to print `Invalid course level:` followed by the error message.
- Use `else` to print a confirmation only when the value is valid.
- Use `finally` to print `Course-level check complete.` every time.

Test it with `abc`, `50`, and `141`.

```python
# Write your function here

def get_course_level():
  # ask for level of the course
  level = input("What is the level of the course?")
  try: # if the input can be converted to an integer
    n = int(level) # convert the input
    if n <= 100 or n >=499: #If the input is outside of the range 100 - 499
      raise ValueError("We require a value that is between 100 and 499") # Send a ValueError message that the number is out of range
  except ValueError as e: # If the input cannot be converted to an integer, print a ValueError message
    print(f"Invalid input detected, {e}")

  else: # If the input was successfully converted to an integer and it was within range, continue
    print("Level confirmed")# print confirmation
  finally: #Last step, no matter what happened with the input
    print("Course-level check complete.") #print closing message

get_course_level()#call function


```

## 6. `range` and a running total (12 minutes)

Write `sum_even_numbers()` with no arguments and no return. Ask the user for a positive integer `n`. Make certain to validate your input using try..except with raise.  Use `for number in range` to calculate and print the sum of all even numbers from 2 through `n`.

Example: if `n` is 8, the total is `2 + 4 + 6 + 8`, or 20.


```python
# Write your function here

def sum_even_numbers():
  #gather input for the integer
  n = input("Kindly suggest a positive integer")
  total = 0 #Define variable total to start with
  try: # if the input can be safely converted to an integer
    n = int(n) #convert input to an integer
    if n < 0: # if the input is not a positive number, raise a value error
      raise ValueError("We need a positive integer") #ValueError we need a positive integer
  except ValueError as e:#if the input cannot be converted to an integer, raise a ValueError
    print(f"Invalid input detected, {e}")
  else:#If the input was safely converted to an integer continue
    '''
    start a loop for the range of zero to the positive integer + 1, to include the acttual integer
    Set it to count every second number, to count and add only the even numbers
    '''
    for number in range(0,n + 1, 2):
      print(f"{number}") #print the even numbers
      total += number # Keep adding each number to the total
    print(f"The total is {total}") #print total

sum_even_numbers()




```

## 8. `continue` and `break` (13 minutes)

Write `collect_scores()` with no arguments and no return. Repeatedly ask for a whole-number score.

- If the user types `done`, end the loop with `break`.
- Convert other entries to integers using `try`.
- If an entry is not an integer, print a message and use `continue`.
- If a score is outside 0–100, print a message and use `continue`.
- For each valid score, add it to `total` and add 1 to `valid_scores`.
- After the loop, print the average if at least one valid score was entered; otherwise print a suitable message.

**Hint:** Because you must check for `done` before conversion, first store the input as text using `.strip().lower()`.


```python
# Write your function here

score = 0 #create variable score
valid_scores = 0 #create variable valid_scores, used to tally up valid scores for the average
def collect_score(): # create function to collect scores
  s = input("Enter a score or type done").strip().lower() #request score via input, make the input lowercase and no spaces
  return(s) #returns clean input

while True: # start indefinite loop
  s = collect_score() # call collect score function
  if s == "done": # first check for done
    break # if we're done, break the loop
  else: # if we're not done
    try: # try conversion
      s = int(s) #convert score into an integer
      if s < 0 or s > 100: # if the score is not within the range of 0-100
        print("This is not a valid score. Please enter a number from 0-100") #print an error message that the score is not in range
        continue # keep going. Discard invalid score, but don't break the loop
      else: # if we have successfully converted the score, keep going
        score += s #Add the score to the variable score
        valid_scores += 1 #record that another score has been added to the list of valid scores. We will use this number to calculate the average.

    except ValueError: #If the score input could not be converted to an integer
      print("This is not a valid entry") # print an error message
      continue #keep going, discard invalid entry, but don't break the loop

if score == 0: #If someone typed done without entering a score, or entered no valid scores
  print("No scores were enetered") #say so
else: #if there were valid scores
  print(f"Your averge is {int(score/valid_scores)}")#print the average. Divide the total scores by the number of valid scores. I made it into a neat integer.







```
