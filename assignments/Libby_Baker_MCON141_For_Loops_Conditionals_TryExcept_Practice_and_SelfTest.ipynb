## 9. `try/except` Inside a `for` Loop

Each item is tried separately. Bad data does not have to crash the whole loop.

**Before running:** trace the values by hand.

````python
entries = ["12", "7", "hello", "20"] #hello is not valid

for entry in entries:
    try:
        number = int(entry)
        print(number, "is a valid integer")
    except ValueError:
        print(entry, "is NOT an integer")
````

## 10. `try/except` + Conditional

First convert safely; then make a decision about the converted number.

**Before running:** trace the values by hand.

````python
entries = ["4", "11", "cat", "18", -2, "4.3", 4.3] #Apparently, you cannot convert a string that is a float directly into an integer

for entry in entries:
    try:
        number = int(entry)
        if number >= 10:
            print(number, "is at least 10") # 11 & 18
        else:
            print(number, "is less than 10") # 4
    except ValueError:
        print(entry, "cannot be converted") # cat
````

## 11. `try/except` + `continue`

If conversion fails, skip the rest of that iteration.

**Before running:** trace the values by hand.

````python
entries = ["5", "oops", "8", "bad", "12", 12.4] #skipping oops and bad

for entry in entries:
    try:
        number = int(entry)
    except ValueError:
        print("Skipping:", entry)
        continue

    print(number, "squared is", number ** 2)
````

## 12. Putting It Together

Convert each value. Skip bad input. Ignore negatives. Stop at 100. Classify the remaining numbers as even or odd.

**Before running:** trace the values by hand.

````python
entries = ["8", "cat", "-3", "11", "24", "100", "6"]

for entry in entries:
    try:
        number = int(entry)
    except ValueError:
        print("Bad input:", entry) #skip cat
        continue

    if number < 0:
        print("Ignoring negative:", number) # skip -3
        continue

    if number == 100:
        print("100 found — stopping") # stop before 100 & 6
        break

    if number % 2 == 0:
        print(number, "even") #8, 24
    else:
        print(number, "odd") #11
````
